```python
import mmcv
from mmseg.apis import init_segmentor, inference_segmentor, show_result_pyplot
import matplotlib.pyplot as plt

config_file = 'path/to/your/custom_cityscapes_config.py'  # Pfad zu deiner Config
checkpoint_file = 'path/to/your/custom_checkpoint.pth'     # Pfad zum Checkpoint
model = init_segmentor(config_file, checkpoint_file, device='cuda:0')
image_path = 'path/to/your/image.png'  # Cityscapes Bild
img = mmcv.imread(image_path)
result = inference_segmentor(model, img)
show_result_pyplot(model, img, result, opacity=0.5)

# Variante B: Maske separat anzeigen (als Klasse pro Pixel)
mask = result[0]
plt.figure(figsize=(8,6))
plt.imshow(mask, cmap='jet', alpha=0.7)
plt.title('Predicted Class Mask')
plt.axis('off')
plt.show()
```
