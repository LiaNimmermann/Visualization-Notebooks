```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("/home/lstracke/blur_preprocessing/blur_preprocessing_git/work_dir_zero_shot/metrics_averaged_over_seed_enhanced_dataset.csv")
df_original = pd.read_csv("/home/lstracke/blur_preprocessing/blur_preprocessing_git/work_dir_zero_shot/metrics_averaged_over_seed_7.0.csv")

label_map = {
    "bl": "Baseline + AllWeatherNet",
    "bw": "Luminance + AllWeatherNet",
    "co": "Color-opponency + AllWeatherNet",
    "sc": "Single-color + AllWeatherNet"
}

label_map_original = {
    "bl": "Baseline",
    "bw": "Luminance",
    "co": "Color-opponency",
    "sc": "Single-color"
}

dataset_map = {
    "acdc_night": "ACDC Night",
    "acdc_fog": "ACDC Fog",
    "acdc_rain": "ACDC Rain",
    "acdc_snow": "ACDC Snow",
    "acdc_full": "ACDC Mean",
    "dark_zurich": "Dark Zurich"
}

# Filter
df = df[(df["Architecture"] == "mask2former") & (df["Sparsity"] == 0.0)]
df_original = df_original[(df_original["Architecture"] == "mask2former") & (df_original["Sparsity"] == 0.0)]

# Map labels
df["Preprocessing"] = df["Preprocessing"].map(label_map)
df_original["Preprocessing"] = df_original["Preprocessing"].map(label_map_original)

df["Dataset"] = df["Dataset"].map(dataset_map)
df_original["Dataset"] = df_original["Dataset"].map(dataset_map)

df_combined = pd.concat([df, df_original], ignore_index=True)

order = [
    "Baseline",
    "Baseline + AllWeatherNet",
    "Luminance",
    "Luminance + AllWeatherNet",
    "Color-opponency",
    "Color-opponency + AllWeatherNet",
    "Single-color",
    "Single-color + AllWeatherNet"
]

dataset_order = ["Dark Zurich", "ACDC Night"]

# Colorblind-safe base colors
cb = sns.color_palette("colorblind", 4)
base_colors = {
    "Baseline": cb[0],
    "Luminance": cb[1],
    "Color-opponency": cb[2],
    "Single-color": cb[3]
}

def base_name(label):
    return label.replace(" + AllWeatherNet", "")

def is_enhanced(label):
    return "AllWeatherNet" in label

# Make sure order is respected
df_combined["Preprocessing"] = pd.Categorical(df_combined["Preprocessing"], categories=order, ordered=True)
df_combined["Dataset"] = pd.Categorical(df_combined["Dataset"], categories=dataset_order, ordered=True)

# Aggregate
stats = (
    df_combined
    .groupby(["Dataset", "Preprocessing"], observed=True)["mIoU"]
    .agg(["mean", "std"])
    .reset_index()
)

# Plot
fig, ax = plt.subplots(figsize=(13, 7))
sns.set_style("whitegrid")

x = np.arange(len(dataset_order))
n_methods = len(order)
width = 0.09
offsets = (np.arange(n_methods) - (n_methods - 1) / 2) * width

for i, method in enumerate(order):
    means = []
    stds = []
    for ds in dataset_order:
        row = stats[(stats["Dataset"] == ds) & (stats["Preprocessing"] == method)]
        if len(row) == 0:
            means.append(np.nan)
            stds.append(0.0)
        else:
            means.append(row["mean"].iloc[0])
            stds.append(0.0 if pd.isna(row["std"].iloc[0]) else row["std"].iloc[0])

    color = base_colors[base_name(method)]
    hatch = "///" if is_enhanced(method) else ""
    edgecolor = "black" if is_enhanced(method) else "none"

    ax.bar(
        x + offsets[i],
        means,
        width=width,
        capsize=3,
        label=method,
        color=color,
        edgecolor=edgecolor,
        hatch=hatch,
        linewidth=1.0
    )

ax.set_title("mIoU across Preprocessing Methods")
ax.set_ylabel("mIoU")
ax.set_xlabel("Dataset")
ax.set_xticks(x)
ax.set_xticklabels(dataset_order)

ax.legend(bbox_to_anchor=(1.02, 0.5), loc="center left", title="Preprocessing")
plt.tight_layout()
plt.show()
```

```python
# Split enhanced vs base
df_base = df_combined[~df_combined["Preprocessing"].str.contains("AllWeatherNet")]
df_enh = df_combined[df_combined["Preprocessing"].str.contains("AllWeatherNet")]

def strip_enh(x):
    return x.replace(" + AllWeatherNet", "")

df_base["Method"] = df_base["Preprocessing"]
df_enh["Method"] = df_enh["Preprocessing"].apply(strip_enh)

# Aggregate means
base_stats = (
    df_base
    .groupby(["Dataset", "Method"])["mIoU"]
    .mean()
    .reset_index()
)

enh_stats = (
    df_enh
    .groupby(["Dataset", "Method"])["mIoU"]
    .mean()
    .reset_index()
)

# Merge to compute delta
merged = pd.merge(
    base_stats,
    enh_stats,
    on=["Dataset", "Method"],
    suffixes=("_base", "_enh")
)

merged["delta"] = merged["mIoU_enh"] - merged["mIoU_base"]

# Plot setup
fig, ax = plt.subplots(figsize=(10, 6))
sns.set_style("whitegrid")

methods = ["Baseline", "Luminance", "Color-opponency", "Single-color"]
dataset_order = ["Dark Zurich", "ACDC Night"]

x = np.arange(len(dataset_order))
width = 0.18
offsets = (np.arange(len(methods)) - (len(methods) - 1) / 2) * width

base_colors = sns.color_palette("colorblind", 4)
color_map = dict(zip(methods, base_colors))

for i, method in enumerate(methods):
    means = []
    for ds in dataset_order:
        row = merged[(merged["Dataset"] == ds) & (merged["Method"] == method)]
        if len(row) == 0:
            means.append(0)
        else:
            means.append(row["delta"].iloc[0])

    ax.bar(
        x + offsets[i],
        means,
        width=width,
        label=method,
        color=color_map[method],
        edgecolor="black"
    )

ax.axhline(0, color="black", linewidth=1)

ax.set_title("mIoU Gain from AllWeatherNet (Enhanced − Base)")
ax.set_ylabel("Δ mIoU")
ax.set_xlabel("Dataset")
ax.set_xticks(x)
ax.set_xticklabels(dataset_order)

ax.legend(title="Preprocessing")
plt.tight_layout()
plt.show()
```
