```python
left_label_width = 40
top_header_height = 40

col_group_gap = 12
row_gap = 12

target_h = 160
scale_h = 14
crop_factor = 0.03

bg = (255,255,255)

# ---------------- Fonts ----------------
font_header = ImageFont.truetype(font_path, 18)
font_label  = ImageFont.truetype(font_path, 18)

def text_size(draw, text, font):
    b = draw.textbbox((0,0), text, font=font)
    return b[2]-b[0], b[3]-b[1]

# ---------------- Loader ----------------
def load_img(path):
    if path.lower().endswith(".pdf"):
        im = convert_from_path(path, dpi=70)[0].convert("RGB")
    else:
        im = Image.open(path).convert("RGB")

    if crop_factor > 0:
        w,h = im.size
        l = int(w*crop_factor)
        t = int(h*crop_factor)
        r = int(w*(1-crop_factor))
        b = int(h*(1-crop_factor))
        if r-l > 5 and b-t > 5:
            im = im.crop((l,t,r,b))

    return im

# ---------------- Referenzbreiten (Cityscapes) ----------------
ref_widths = {}
for col in columns:
    ws = []
    for f in file_map[("Cityscapes", col)]:
        im = load_img(f)
        w,h = im.size
        ws.append(int(w * target_h / h))
    ref_widths[col] = ws

# ---------------- Cells bauen ----------------
cells = {}
col_widths = {}
row_heights = {}

for row_name,_ in rows:
    heights=[]
    for col in columns:
        parts=[]
        files = file_map.get((row_name,col), [])
        for i,f in enumerate(files):
            im = load_img(f)
            w = ref_widths[col][i] if i < len(ref_widths[col]) else ref_widths[col][-1]

            if row_name == "Skala":
                im = im.resize((w, scale_h), Image.LANCZOS)
            else:
                im = im.resize((w, target_h), Image.LANCZOS)

            parts.append(im)

        if not parts:
            continue

        total_w = sum(p.width for p in parts)
        h = max(p.height for p in parts)
        cell = Image.new("RGB",(total_w,h),bg)
        x=0
        for p in parts:
            cell.paste(p,(x,(h-p.height)//2))
            x += p.width

        cells[(row_name,col)] = cell
        col_widths[col] = max(col_widths.get(col,0), total_w)
        heights.append(h)
    row_heights[row_name] = max(heights)

# ---------------- Canvas Size ----------------
#W = left_label_width + sum(col_widths[c] for c in columns) + col_group_gap*(len(columns)-1)
# Nach Erstellen aller Zellen:
W = left_label_width + max(sum(cells[(row_name,col)].width for col in columns) for row_name,_ in rows) + col_group_gap*(len(columns)-1)
H = top_header_height + sum(row_heights.values()) + row_gap*(len(rows)-1)

canvas = Image.new("RGB",(W,H),bg)
draw = ImageDraw.Draw(canvas)

# ---------------- Header ----------------
x = left_label_width
for col in columns:
    w = col_widths[col]
    tw,th = text_size(draw,col,font_header)
    draw.text((x+(w-tw)/2,6), col, font=font_header, fill=(0,0,0))
    x += w + col_group_gap

# ---------------- Rows ----------------
y = top_header_height
for row_name,_ in rows:
    h = row_heights[row_name]

    # Label direkt auf Canvas, vertikal (wie obere Header)
    # Label direkt als Image erzeugen, dann drehen
    if row_name != "Skala":
        # Textgröße
        tw, th = text_size(draw, row_name, font_label)
        pad = 6  # genug Platz oben/unten für Rotation
        temp = Image.new("RGBA", (tw + pad*2, th + pad*2), (0,0,0,0))
        td = ImageDraw.Draw(temp)
        td.text((pad, pad), row_name, font=font_label, fill=(0,0,0))
        temp = temp.rotate(90, expand=True)

        
        cx = left_label_width // 2
        cy = y + row_heights[row_name] // 2
        canvas.paste(temp, (cx - temp.width//2 - 4, cy - temp.height//2), temp)





    # Zellen spaltenweise
    x = left_label_width
    for col in columns:
        c = cells[(row_name,col)]
        canvas.paste(c,(x,y+(h-c.height)//2))
        x += col_widths[col] + col_group_gap

    y += h + row_gap

# ---------------- Save ----------------
canvas.save("teaser_combined.png")
canvas.save("teaser_combined.pdf", "PDF", resolution=300)

display(canvas)
print("fertig")
```
